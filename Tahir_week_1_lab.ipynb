```python
#Q2
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
print (is_prime(5))
```

```python
def fibonacci(n):
    seq = []
    x, y = 0, 1
    for _ in range(n):
        seq.append(x)
        x, y = y, x + y
    return seq

print("Q3:", fibonacci(10))
```

```python
def remove_duplicates(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

print("Q4:", remove_duplicates([1, 2, 2, 3, 1, 4, 3]))  # [1, 2, 3, 4]

```

```python
def count_chars(s):
    vowels = consonants = digits = spaces = 0
    for ch in s:
        if ch.isalpha():
            if ch.lower() in "aeiou":
                vowels += 1
            else:
                consonants += 1
        elif ch.isdigit():
            digits += 1
        elif ch.isspace():
            spaces += 1
    return vowels, consonants, digits, spaces

print("Q5:", count_chars("Hello World 123"))
```

```python
def multiply(*args):
    result = 1
    for num in args:
        result *= num
    return result

print("Q6:", multiply(2, 3, 4))
```

```python
arr20 = np.arange(1, 17).reshape(4, 4).astype(float)
arr20 += np.eye(4)  # nudge so it's invertible (raw 1-16 matrix is singular)
det20 = np.linalg.det(arr20)
print("Q20 transpose:\n", arr20.T)
print("Q20 determinant:", det20)
if abs(det20) > 1e-10:
    print("Q20 inverse:\n", np.linalg.inv(arr20))
else:
    print("Q20: matrix is singular, no inverse")

```
